# Airbnb price analytics: project structure

The study estimates nightly advertised listing prices in GBP using London and Bristol development snapshots and an unseen Greater Manchester transfer evaluation.

| Component | Responsibility | Status |
| --- | --- | --- |
| Exploratory analysis | Person 1: provenance, data quality, distributions and leakage audit | Executed successfully |
| Lakehouse and ETL | Person 2: immutable Bronze inputs, typed Silver records, Gold interfaces and city-parameterized ingestion | Implementation pending |
| Price model | Person 3: training-only baseline, model selection, held-out MAE and Gold predictions | Implementation pending |
| Dashboard and Genie | Person 4: city overview, cross-city comparison and model diagnostics | Implementation pending |

## Methodological requirements

Preserve raw files and source checksums. Reserve per-city test sets before model selection; fit preprocessing and baseline statistics on training data only. Exclude price quotations and revenue-derived predictors.

Greater Manchester data remains inaccessible until successful A/B pipeline and model versions are frozen. Record the unchanged first-run result before making corrections.

Source code, contracts and aggregate evidence are maintained in the repository. Execution parameters and storage interfaces are defined in `docs/setup_databricks.md` and `docs/data_contract.md`.
