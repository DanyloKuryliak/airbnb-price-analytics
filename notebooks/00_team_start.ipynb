Repository: https://github.com/DanyloKuryliak/airbnb-price-analytics

## Data and namespaces

Currency: GBP. London snapshot: 2026-09-17. Bristol snapshot: 2026-09-22. Detailed immutable A/B inputs and the checksum manifest are in `/Volumes/workspace/airbnb_eda/raw/`. EDA aggregate artifacts are in `/Volumes/workspace/airbnb_eda/artifacts/`.

Shared lakehouse namespaces: `workspace.airbnb_bronze`, `workspace.airbnb_silver`, `workspace.airbnb_gold`. These are reserved schemas; production tables are not implemented. Agree column contracts before creating dependent tables.

## Execution and evaluation

Run A/B EDA, agree cleaning contracts, implement ETL, then reserve A/B test sets before model selection. Fit preprocessing and room-type/accommodates median baselines on training rows only. Report per-city and combined test MAE in GBP.

Greater Manchester is locked. Do not acquire, inspect or query its data before successful A/B pipeline/model versions are frozen at the end of Week 1. Run those artifacts unchanged for the first third-city evaluation and save evidence before fixes.
